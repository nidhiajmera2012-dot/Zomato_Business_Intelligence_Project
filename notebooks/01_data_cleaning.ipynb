```python
import pandas as pd
import numpy as np
import sys

sys.path.append("../src")

data_folder = "../data/cleaned/"
print("Ready")
```

```python
names = ["orders", "order_items", "customers", "restaurants", "menu",
         "delivery_partners", "payments", "customer_feedback",
         "promotions", "cities", "traffic", "weather"]

tables = {}

for n in names:
    tables[n] = pd.read_csv(data_folder + n + ".csv")
    print(n, tables[n].shape)
```

```python
for n in names:
    df = tables[n]
    print(n, "| missing values:", df.isnull().sum().sum(), "| duplicate rows:", df.duplicated().sum())
```

```python
import clean

clean_tables = {}

clean_tables["orders"] = clean.clean_orders(tables["orders"])
clean_tables["order_items"] = clean.clean_order_items(tables["order_items"])
clean_tables["customers"] = clean.clean_customers(tables["customers"])
clean_tables["restaurants"] = clean.clean_restaurants(tables["restaurants"])
clean_tables["menu"] = clean.clean_menu(tables["menu"])
clean_tables["delivery_partners"] = clean.clean_delivery_partners(tables["delivery_partners"])
clean_tables["payments"] = clean.clean_payments(tables["payments"])
clean_tables["customer_feedback"] = clean.clean_customer_feedback(tables["customer_feedback"])
clean_tables["promotions"] = clean.clean_promotions(tables["promotions"])
clean_tables["cities"] = clean.clean_cities(tables["cities"])
clean_tables["traffic"] = clean.clean_traffic(tables["traffic"])
clean_tables["weather"] = clean.clean_weather(tables["weather"])

for n in names:
    print(n, tables[n].shape, "->", clean_tables[n].shape)
```

```python
date_cols = [
    ("orders", "OrderDate"),
    ("customers", "RegistrationDate"),
    ("delivery_partners", "JoiningDate"),
    ("payments", "PaymentDate"),
    ("promotions", "StartDate"),
    ("promotions", "EndDate"),
    ("traffic", "Date"),
    ("weather", "Date"),
]

time_cols = [
    ("orders", "OrderTime"),
    ("traffic", "Time"),
    ("restaurants", "OpeningTime"),
    ("restaurants", "ClosingTime"),
]

print("DATE COLUMNS (expected pattern: 2023-06-17)")
for t, col in date_cols:
    s = tables[t][col].dropna().astype(str)
    bad = (~s.str.match(r"^\d{4}-\d{2}-\d{2}$")).sum()
    print(t, col, "| sample:", s.iloc[0], "| not matching pattern:", bad, "| missing:", tables[t][col].isnull().sum())

print()
print("TIME COLUMNS (expected pattern: 11:51:00)")
for t, col in time_cols:
    s = tables[t][col].dropna().astype(str)
    bad = (~s.str.match(r"^\d{2}:\d{2}:\d{2}$")).sum()
    print(t, col, "| sample:", s.iloc[0], "| not matching pattern:", bad, "| missing:", tables[t][col].isnull().sum())
```
